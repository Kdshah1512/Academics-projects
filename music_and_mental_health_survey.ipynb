The client aims to use data on user behavior, coupled with mental health, for real user engagement and personalized marketing. Load dataset and required libraries

**cleaning data and fill missing values**








```python

mean_age = data['Age'].mean()
data['Age'].fillna(mean_age, inplace=True)
data['Primary streaming service'].fillna(data['Primary streaming service'].mode()[0], inplace=True)
data['While working'].fillna(data['While working'].mode()[0], inplace=True)
data['Instrumentalist'].fillna(data['Instrumentalist'].mode()[0], inplace=True)
mean_bpm = data['BPM'].mean()
data['BPM'].fillna(mean_bpm, inplace=True)
data['Music effects'].fillna(data['Music effects'].mode()[0], inplace=True)
data['Composer'].fillna(data['Composer'].mode()[0], inplace=True)
data['Foreign languages'].fillna(data['Foreign languages'].mode()[0], inplace=True)

print(data.isnull().sum())
```

**Exploratory Data Analysis**

***Business Questions and Analysis***

Question 1:How does music genre preference vary across different age groups?

```python
age_bins = [0, 18, 25, 35, 50, 100]
age_labels = ['<18', '18-25', '26-35', '36-50', '50+']
data['Age Group'] = pd.cut(data['Age'], bins=age_bins, labels=age_labels, right=False)

genre_by_age = pd.crosstab(data['Age Group'], data['Fav genre'], normalize='index')
genre_by_age.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Music Genre Preference by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Proportion')
plt.legend(title='Genre', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

**Answer: In above result showing Musical evolution accompanies aging , as preferences migrate from popular pop and hip hop to more refined and varied genres like rock jazz and classical**

question 2:Is there a relationship between daily listening hours and self-reported mental health scores?

```python
print("\n2) Is there a relationship between daily listening hours and self-reported mental health scores?")
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
mental_health_vars = ['Anxiety', 'Depression', 'Insomnia']

for i, var in enumerate(mental_health_vars):
    sns.scatterplot(data=data, x='Hours per day', y=var, ax=axes[i])
    axes[i].set_title(f'Hours per Day vs {var}')

plt.tight_layout()
plt.show()
correlations = data[['Hours per day'] + mental_health_vars].corr()['Hours per day'].drop('Hours per day')
print("Correlations with listening hours:")
print(correlations)
```

**Answer: Correlation analysis will determine whether a positive or negative correlation exists between daily music listening habits and mental health metrics, such as anxiety, depression, and insomnia, with scatter plots illustrating these associations.**

question :3 Which streaming services are most popular among users reporting high levels of anxiety or depression?

```python
print("\n3) Which streaming services are most popular among users reporting high levels of anxiety or depression?")
def high_score(x):
    return x > x.median()

high_anxiety = data[high_score(data['Anxiety'])]
high_depression = data[high_score(data['Depression'])]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

high_anxiety['Primary streaming service'].value_counts().plot(kind='pie', ax=ax1, autopct='%1.1f%%')
ax1.set_title('Streaming Services for High Anxiety Users')

high_depression['Primary streaming service'].value_counts().plot(kind='pie', ax=ax2, autopct='%1.1f%%')
ax2.set_title('Streaming Services for High Depression Users')

plt.tight_layout()
plt.show()
```

**answer: Pie charts  provides a graphical representation of the most frequently used streaming services among individuals with anxiety and depression, shedding light on their affinity for specific platform**

Question 4 :How does the frequency of music exploration trying new genres ,artists relate to mental health scores?

```python
print("\n4) How does the frequency of music exploration (trying new genres/artists) relate to mental health scores?")
data['Exploratory'] = data['Exploratory'].map({'Yes': 1, 'No': 0})
data['Foreign languages'] = data['Foreign languages'].map({'Yes': 1, 'No': 0})
exploration_vars = ['Exploratory', 'Foreign languages']
mental_health_vars = ['Anxiety', 'Depression', 'Insomnia']

correlations = data[exploration_vars + mental_health_vars].corr().loc[exploration_vars, mental_health_vars]

plt.figure(figsize=(10, 6))
sns.heatmap(correlations, annot=True, cmap='coolwarm')
plt.title('Correlation: Music Exploration vs Mental Health Scores')
plt.show()
```

**Answer: It would provide insight into how music exploratory behavior relates to mental health scores and answer the question of whether such a tendency to switch genres and artists correlates positively with high or low levels of anxiety, depression, and insomnia. Negative correlations, however, might well suggest a possible benefit on mental health.**

Question 5 : Is there a difference in listening habits between users who use music to improve their mood versus those who don't?

```python
print("\n5) Is there a difference in listening habits between users who use music to improve their mood versus those who don't?")
mood_improvement = data['Music effects'].str.contains('Improve', case=False, na=False)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

sns.boxplot(x=mood_improvement, y='Hours per day', data=data, ax=ax1)

ax1.set_title('Daily Listening Hours: Mood Improvement vs No Mood Improvement')
ax1.set_xlabel('Uses Music for Mood Improvement')

sns.boxplot(x=mood_improvement, y='Instrumentalist', data=data, ax=ax2)
ax2.set_title('Instrumentalist Score: Mood Improvement vs No Mood Improvement')
ax2.set_xlabel('Uses Music for Mood Improvement')

plt.tight_layout()
plt.show()
```

**Answer: A comparative analysis using boxplots will investigate differences in daily music listening and instrumentalist scores among mood-enhancing music users and non-users**
